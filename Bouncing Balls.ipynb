```python
import pygame 
import random
pygame.init()

width, height = 800, 600
screen = pygame.display.set_mode((width, height))
clock = pygame.time.Clock()

white = (255, 255, 255)
black = (0, 0, 0)
red = (255, 0, 0)
green = (0, 255, 0)
blue = (0, 0, 255)
```

```python
ball_radius = 20

class Ball:
    def __init__(self, color):
        self.x = random.randint(ball_radius, width - ball_radius)
        self.y = random.randint(ball_radius, height - ball_radius)
        self.vx = random.randint(2,6)
        self.vy = random.randint(-6,-2)
        self.color = color
        
    def move(self):
        self.x += self.vx
        self.y += self.vy
        
        if self.x <= ball_radius or self.x >= width - ball_radius:
            self.vx *= -1
            if self.x <= ball_radius:
                self.x = ball_radius
            elif self.x >= width - ball_radius:
                self.x = width - ball_radius
        if self.y <= ball_radius or self.y >= height - ball_radius:
            self.vy *= -1
            if self.y <= ball_radius:
                self.y = ball_radius
            elif self.y >= height - ball_radius:
                self.y = height - ball_radius
    def draw(self):
        pygame.draw.circle(screen, self.color, (self.x, self.y), ball_radius)


```

```python
#balls = [Ball((random.randint(0,255), random.randint(0,255), random.randint(0,255))) for _ in range(10)]
balls = [Ball(blue), Ball(red)]
```

```python
running = True
while running:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False
            
    screen.fill(white)
    
    for ball in balls:
        ball.move()
        ball.draw()
    
    pygame.display.flip()
    clock.tick(60)
    
pygame.quit()
```
